## Architecture (v3 — Full Failure Analysis Applied + Autoencoder Added)

### Four-Layer Ensemble
| Layer | Model | Role | Training Data |
|---|---|---|---|
| 1a | Isolation Forest | Normal manifold score (feature for XGBoost) | Normal-only |
| 1b | **Autoencoder (NEW)** | Reconstruction error anomaly score | Normal-only |
| 2 | XGBoost | Primary backbone classifier (binary) | Full labelled dataset |
| 3 | CNN-BiLSTM + Attention | Temporal refiner on flagged windows | Full dataset |

### v3 vs v2 Changes (from complete_failure_analysis.md)
| Issue | v2 | v3 Fix |
|---|---|---|
| **Autoencoder missing** | Not implemented | **Added: 128→64→32→64→128 PyTorch AE** |
| IF used as voting layer (F1=0.09) | IF binary vote | IF score → XGBoost feature (augmented) |
| IF FPR target too tight | 0.03 | 0.08 (better recall trade-off) |
| XGBoost SMOTE fake frozen samples | BorderlineSMOTE | `scale_pos_weight` (research report plan) |
| XGBoost binary:logistic discarded | multi:softprob | `binary:logistic` + `eval_metric=aucpr` |
| XGB early stop at 72/500 | max_depth=7, lr=0.05 | max_depth=5, lr=0.03, n_estimators=800 |
| CNN total collapse (F1=0.06) | Adam lr=1e-3 | Adam lr=3e-4, clipnorm=1.0 |
| CNN EarlyStopping fired at epoch 9 | patience=8, val_loss | patience=15, val_auc (correct metric) |
| CNN ReduceLR wrong monitor | val_loss | val_auc, mode=max, patience=5 |
| CNN epochs too low | 50 | 80 |
| Ensemble majority vote (F1=0.14) | 3-way majority | Weighted soft score + calibrated threshold |
| Thresholds hardcoded | XGB=0.70, CNN=0.50 | All calibrated on val F1 |

## Cell 3 — Memory-Safe Chunked Data Loading

````python
#cell 3

BINARY_COLS = {
    "P_101","P_102","P_201","P_202","P_203","P_204","P_205","P_206",
    "P_301","P_302","P_401","P_402","P_403","P_404","P_501","P_502",
    "P_601","P_602","P_603","UV_401",
    "MV_101","MV_201","MV_301","MV_302","MV_303","MV_304",
    "UF_Backwash_Active","RO_Cleaning_Active","Chemical_Low_Alarm",
    "High_Fouling_Alarm","High_Level_Alarm","High_Pressure_Alarm",
    "System_Run","Energy_Monitor_Enable",
}
DROP_COLS = {
    "Energy_P101","Energy_P301","Energy_P501","Energy_Total",
    "PIT_503",   # r=1.0 with PIT_501
    "AIT_203",   # r=1.0 with AIT_402
}


def downcast_chunk(chunk):
    chunk.drop(columns=[c for c in DROP_COLS if c in chunk.columns], inplace=True)
    for col in chunk.columns:
        if col in ("Timestamp","ATTACK_NAME","MITRE_ID","ATTACK_ID"): continue
        if chunk[col].dtype == "bool": chunk[col] = chunk[col].astype(np.int8)
        elif col in BINARY_COLS: chunk[col] = chunk[col].fillna(0).astype(np.int8)
        elif chunk[col].dtype in ("int64","int32"): chunk[col] = chunk[col].astype(np.int32)
        elif chunk[col].dtype == "float64": chunk[col] = chunk[col].astype(np.float32)
    return chunk


def load_dataset(path, label):
    print(f"Loading {label}: {path}")
    t0, chunks = time.time(), []
    for i, chunk in enumerate(pd.read_csv(path, chunksize=CHUNKSIZE, low_memory=False)):
        chunks.append(downcast_chunk(chunk))
        if (i+1) % 5 == 0: print(f"  {sum(len(c) for c in chunks):,} rows...", end="\r")
    df = pd.concat(chunks, ignore_index=True)
    del chunks; gc.collect()
    for tc in ["Timestamp","timestamp","Time","time"]:
        if tc in df.columns:
            df[tc] = pd.to_datetime(df[tc], format="mixed", utc=True, errors="coerce")
            df = df.dropna(subset=[tc]).sort_values(tc).reset_index(drop=True)
            df.rename(columns={tc:"Timestamp"}, inplace=True); break
    else:
        df.insert(0,"Timestamp",pd.date_range("2026-01-01",periods=len(df),freq="100ms"))
    assert "ATTACK_ID" in df.columns, "ATTACK_ID column missing!"
    df["ATTACK_ID"] = df["ATTACK_ID"].fillna(0).astype(np.int32)
    if df.isnull().sum().sum(): df.ffill(inplace=True); df.bfill(inplace=True)
    mb = df.memory_usage(deep=True).sum() / 1024**2
    print(f"  {df.shape[0]:,} rows x {df.shape[1]} cols | {mb:.0f} MB | {time.time()-t0:.1f}s")
    return df


# ── Load both datasets ────────────────────────────────────────────────────────
# Dataset 1 (NORMAL_PATH)  : pure normal operation — full 24h, large
# Dataset 2 (ATTACK_PATH)  : normal + attack mixed — normal rows are very few
#
# Strategy:
#   IF / AE / Scaler  → fit on Dataset 1 normal rows only (clean, large)
#   XGBoost / CNN     → Dataset 1 normal rows  +  Dataset 2 attack rows only
#                        (Dataset 2 normal rows excluded — already covered by D1)
#
# This gives a realistic normal/attack ratio and avoids the 86.6% attack skew
# that caused CNN collapse in v2.

df_normal = load_dataset(NORMAL_PATH, "Normal-24h (pure)")
df_attack = load_dataset(ATTACK_PATH, "Attack-24h (normal+attack)")

# Ensure df_normal contains only normal rows
if (df_normal["ATTACK_ID"] != 0).any():
    n_before = len(df_normal)
    df_normal = df_normal[df_normal["ATTACK_ID"] == 0].reset_index(drop=True)
    print(f"  Dropped {n_before - len(df_normal):,} non-normal rows from normal dataset")

# Separate attack dataset into attack-only rows
# (the very few normal rows in attack CSV are already in df_normal — skip them)
df_attack_only = df_attack[df_attack["ATTACK_ID"] != 0].copy().reset_index(drop=True)
df_attack_only["attack_type"] = df_attack_only["ATTACK_ID"].map(ATTACK_TYPE_MAP).fillna("unknown")

print(f"\n{'─'*55}")
print(f"Normal dataset (Dataset 1)  : {len(df_normal):,} rows  (all ATTACK_ID=0)")
print(f"Attack dataset (Dataset 2)  : {len(df_attack):,} rows total")
print(f"  → attack-only rows used   : {len(df_attack_only):,}")
normal_in_atk = (df_attack["ATTACK_ID"] == 0).sum()
print(f"  → normal rows in D2 (discarded, covered by D1): {normal_in_atk:,}")
print(f"{'─'*55}")
atk_dist = df_attack_only["ATTACK_ID"].value_counts().sort_index()
for aid, cnt in atk_dist.items():
    print(f"  ATTACK_ID={aid:<3d} ({ATTACK_TYPE_MAP.get(int(aid),'?'):<22s}): {cnt:,}")
del df_attack; gc.collect()
````

## Cell 4 — Feature Engineering (function definition only)

`engineer_features()` is defined here and called in Cell 5 after the split,
so that `dt_stats` (timing statistics) are computed from the normal dataset
and reused without leakage when applied to the attack dataset.

All 9 feature groups from research report:
1. Temporal (delta_t, jitter, delay flag)
2. Physical mass balance (flow diffs, level diffs, DPIT second derivative)
3. Control/PLC rules (pump-flow, pH-pump inconsistency, valve mismatch)
4. MITRE-aligned physics inconsistency flags (6 binary signatures)
5. Pump duty cycles (P_203 60s, P_403 30s)
6. Sensor rate-of-change
7. Rolling statistics (mean, std, z-score)
8. Lag features ([1,5,10,30,100] steps for 14 key sensors)
9. Replay frozen-sensor detection (multi-sensor rolling std collapse)


````python
#cell 4
EXCL = {"Timestamp","ATTACK_ID","ATTACK_NAME","MITRE_ID","attack_type"}


def engineer_features(df, dt_stats=None):
    """9-group feature engineering. In-place. Returns (df, feature_cols, dt_stats)."""
    sensor_cols = [c for c in df.columns if any(c.startswith(p)
                   for p in ["LIT","PIT","AIT","DPIT","FIT"])]

    # ── Group 1: Temporal ─────────────────────────────────────────────────────
    if pd.api.types.is_datetime64_any_dtype(df.get("Timestamp", pd.Series())):
        df["delta_t"] = df["Timestamp"].diff().dt.total_seconds().fillna(EXPECTED_DT).astype(np.float32)
    else:
        df["delta_t"] = np.float32(EXPECTED_DT)
    if dt_stats is None:
        dt_mean = float(df["delta_t"].mean())
        dt_std  = float(df["delta_t"].std()) + 1e-9
    else:
        dt_mean, dt_std = dt_stats["dt_mean"], dt_stats["dt_std"]
    df["dt_z"]    = ((df["delta_t"] - dt_mean) / dt_std).astype(np.float32)
    df["dt_rm50"] = df["delta_t"].rolling(50, min_periods=1).mean().astype(np.float32)
    df["jitter"]  = (df["delta_t"] - df["dt_rm50"]).abs().astype(np.float32)
    df["delay_f"] = (df["delta_t"] > 0.2).astype(np.int8)

    # ── Group 2: Physical mass balance ────────────────────────────────────────
    for a, b in [("LIT_101","LIT_301"),("LIT_301","LIT_401")]:
        if a in df.columns and b in df.columns:
            df[f"ldiff_{a}_{b}"] = (df[a] - df[b]).astype(np.float32)
    for a, b in [("FIT_101","FIT_201"),("FIT_201","FIT_301"),
                 ("FIT_301","FIT_401"),("FIT_401","FIT_501")]:
        if a in df.columns and b in df.columns:
            df[f"fbal_{a}_{b}"] = (df[a] - df[b]).astype(np.float32)
    if "DPIT_301" in df.columns:
        df["dpit_roc"]  = df["DPIT_301"].diff().fillna(0).astype(np.float32)
        df["dpit_roc2"] = df["dpit_roc"].diff().fillna(0).astype(np.float32)
    if "PIT_501" in df.columns:
        df["pressure_spike"] = (df["PIT_501"] > 1800).astype(np.int8)
    for pc in ["AIT_202","AIT_402"]:
        if pc in df.columns:
            df[f"{pc}_dev"] = ((df[pc] < 650) | (df[pc] > 850)).astype(np.int8)

    # ── Group 3: Control / PLC rules ──────────────────────────────────────────
    if "P_101" in df.columns and "FIT_101" in df.columns:
        thr = max(float(df["FIT_101"].quantile(0.10)), 0.05)
        df["pump_flow_inc"] = ((df["P_101"]==1) & (df["FIT_101"] < thr)).astype(np.int8)
    if "AIT_202" in df.columns and "P_203" in df.columns:
        df["ph_pump_inc"] = ((df["AIT_202"] > 750) & (df["P_203"] == 0)).astype(np.int8)
    for mv, fit in [("MV_101","FIT_101"),("MV_301","FIT_301")]:
        if mv in df.columns and fit in df.columns:
            thr = max(float(df[fit].quantile(0.05)) * 2, 0.1)
            df[f"vmm_{mv}"] = ((df[mv] == 0) & (df[fit] > thr)).astype(np.int8)
    for pc in ["P_101","P_301","P_501"]:
        if pc in df.columns:
            df[f"{pc}_d60"] = df[pc].rolling(60, min_periods=1).mean().astype(np.float32)

    # ── Group 4: MITRE-aligned physics inconsistency flags ────────────────────
    # Valve Manipulation: pump ON but no flow (T0849)
    if all(c in df.columns for c in ["P_101","FIT_101"]):
        df["pump_on_zero_flow"] = ((df["P_101"]==1) & (df["FIT_101"] < 0.2)).astype(np.int8)
    # Tank Overflow: pumps off, inlet open, level high (T0816)
    if all(c in df.columns for c in ["P_101","P_102","MV_101","LIT_101"]):
        df["overflow_pattern"] = (
            (df["P_101"]==0) & (df["P_102"]==0) &
            (df["MV_101"]==1) & (df["LIT_101"] > 700)
        ).astype(np.int8)
    # pH Manipulation: low pH but acid pump OFF (T0831)
    if all(c in df.columns for c in ["AIT_202","P_203"]):
        df["low_ph_no_acid"] = ((df["AIT_202"] < 550) & (df["P_203"] == 0)).astype(np.int8)
    # Chemical Depletion: all 4 dosing pumps ON simultaneously (T0836)
    dp = ["P_203","P_205","P_206","P_403"]
    if all(c in df.columns for c in dp):
        df["all_dosing_on"] = (df[dp].sum(axis=1) == 4).astype(np.int8)
    # Membrane Damage: DPIT rising without backwash (T0830)
    if all(c in df.columns for c in ["DPIT_301","UF_Backwash_Active"]):
        df["membrane_attack_sig"] = (
            (df["DPIT_301"] > 60) & (df["UF_Backwash_Active"] == 0)
        ).astype(np.int8)
    # Replay: multi-sensor frozen (rolling std collapse) (T0843)
    std_s = [s for s in ["LIT_101","AIT_202","FIT_101","DPIT_301"] if s in df.columns]
    if len(std_s) >= 3:
        for s in std_s:
            df[f"{s}_rs100"] = df[s].rolling(100, min_periods=10).std().fillna(0).astype(np.float32)
        rs_cols = [f"{s}_rs100" for s in std_s]
        df["multi_frozen"] = (df[rs_cols] < 0.5).all(axis=1).astype(np.int8)

    # ── Group 5: Pump duty cycles ─────────────────────────────────────────────
    if "P_203" in df.columns:
        df["p203_duty_60s"] = df["P_203"].rolling(600, min_periods=1).mean().astype(np.float32)
        df["p203_duty_20s"] = df["P_203"].rolling(200, min_periods=1).mean().astype(np.float32)
    if "P_403" in df.columns:
        df["p403_duty_30s"] = df["P_403"].rolling(300, min_periods=1).mean().astype(np.float32)
        df["p403_duty_10s"] = df["P_403"].rolling(100, min_periods=1).mean().astype(np.float32)
    if "P_205" in df.columns:
        df["p205_duty_60s"] = df["P_205"].rolling(600, min_periods=1).mean().astype(np.float32)
    if "P_101" in df.columns:
        df["p101_duty_200"] = df["P_101"].rolling(200, min_periods=1).mean().astype(np.float32)

    # ── Group 6: Sensor rate of change ────────────────────────────────────────
    for sc in sensor_cols:
        df[f"{sc}_roc"] = df[sc].diff().fillna(0).astype(np.float32)

    # ── Group 7: Rolling statistics (mean, std, z-score) ─────────────────────
    for sc in sensor_cols:
        rm = df[sc].rolling(ROLL_W, min_periods=1).mean().astype(np.float32)
        rs = df[sc].rolling(ROLL_W, min_periods=1).std().fillna(0).astype(np.float32)
        df[f"{sc}_rm"] = rm
        df[f"{sc}_rs"] = rs
        df[f"{sc}_rz"] = ((df[sc] - rm) / rs.replace(0, 1e-9)).astype(np.float32)

    # C-3 fix: Chlorine_Residual replacement (75% near-zero — noise)
    if "Chlorine_Residual" in df.columns:
        df["cl_res_roc"] = df["Chlorine_Residual"].diff().fillna(0).astype(np.float32)
    if "Chlorine_Tank_Level" in df.columns:
        df["cl_tank_drain"] = df["Chlorine_Tank_Level"].diff().fillna(0).astype(np.float32)
    drain_cols = [c for c in ["Acid_Tank_Level","Chlorine_Tank_Level","Coagulant_Tank_Level"]
                  if c in df.columns]
    if len(drain_cols) == 3:
        drains = [df[c].diff().fillna(0) for c in drain_cols]
        df["all_tanks_drain"] = (
            (drains[0] < -0.03) & (drains[1] < -0.03) & (drains[2] < -0.03)
        ).astype(np.int8)

    # ── Group 8 & 9: Lag features ─────────────────────────────────────────────
    for col in [c for c in LAG_COLS if c in df.columns]:
        for lag in LAG_STEPS:
            df[f"{col}_L{lag}"] = df[col].shift(lag).bfill().astype(np.float32)

    # ── Feature selection: drop near-zero-variance + Chlorine_Residual noise ──
    numeric_types = (np.float32, np.float64, np.int32, np.int16, np.int8, np.int64)
    candidates = [c for c in df.columns
                  if c not in EXCL and df[c].dtype.type in numeric_types]
    low_var   = [c for c in candidates if df[c].std() < 0.01]
    cl_noise  = [c for c in candidates if "Chlorine_Residual" in c and not c.endswith("_roc")]
    feat_cols = [c for c in candidates if c not in low_var and c not in cl_noise]
    print(f"  {len(candidates)} candidates → {len(low_var)} low-var + {len(cl_noise)} Cl-noise dropped → {len(feat_cols)} final")
    return df, feat_cols, {"dt_mean": dt_mean, "dt_std": dt_std}

````

## Cell 5 — Labels, Split & Scaling

````python
#cell 5
# =============================================================================
# SPLIT STRATEGY: Use BOTH datasets properly
#
# Normal data  → from df_normal (full 24h, large, clean)
# Attack data  → from df_attack_only (attack rows only from Dataset 2)
#
# Each is split 70/15/15 independently then concatenated per split.
# This ensures:
#   - Normal manifold is fully represented (not starved at 13% like before)
#   - Attack types are properly represented per split
#   - No temporal leakage between train/val/test within each dataset
# =============================================================================

le = LabelEncoder()
le.fit(df_attack_only["ATTACK_ID"])
n_classes      = len(le.classes_)
normal_encoded = int(le.transform([0])[0]) if 0 in le.classes_ else -1

# ── Feature engineering (applied to both DataFrames) ─────────────────────────
print("Engineering features — normal dataset...")
df_normal, feat_n, dt_stats = engineer_features(df_normal)
print("\nEngineering features — attack-only dataset...")
df_attack_only, feat_a, _   = engineer_features(df_attack_only, dt_stats=dt_stats)

# Union of features (fill any missing in normal with 0)
feature_cols = feat_a
for c in feature_cols:
    if c not in df_normal.columns:
        df_normal[c] = np.float32(0.0)
gc.collect()
print(f"\nFeature count: {len(feature_cols)}")

# ── Split each dataset 70/15/15 chronologically ──────────────────────────────
def chrono_split(df, feature_cols, label_col="ATTACK_ID"):
    N   = len(df)
    n_tr  = int(N * TRAIN_FRAC)
    n_val = int(N * VAL_FRAC)
    X     = df[feature_cols].values.astype(np.float32)
    y_bin = (df[label_col].values > 0).astype(np.int8)
    y_id  = df[label_col].values.astype(np.int32)
    return (X[:n_tr],          y_bin[:n_tr],          y_id[:n_tr],
            X[n_tr:n_tr+n_val], y_bin[n_tr:n_tr+n_val], y_id[n_tr:n_tr+n_val],
            X[n_tr+n_val:],     y_bin[n_tr+n_val:],     y_id[n_tr+n_val:])

(X_nor_tr, y_nor_bin_tr, y_nor_id_tr,
 X_nor_va, y_nor_bin_va, y_nor_id_va,
 X_nor_te, y_nor_bin_te, y_nor_id_te) = chrono_split(df_normal, feature_cols)

(X_atk_tr, y_atk_bin_tr, y_atk_id_tr,
 X_atk_va, y_atk_bin_va, y_atk_id_va,
 X_atk_te, y_atk_bin_te, y_atk_id_te) = chrono_split(df_attack_only, feature_cols)

# Keep pure normal train for IF/AE/Scaler fitting (before combining)
X_nor_train_raw = X_nor_tr.copy()

# ── Combine for supervised training ──────────────────────────────────────────
# Normal (from D1) + Attack (from D2) — interleave by concatenating then sorting
# For tabular models (XGBoost): shuffle is fine
# For CNN: we keep separate arrays and build separate loaders that WeightedSampler handles

X_train_raw = np.concatenate([X_nor_tr, X_atk_tr], axis=0)
X_val_raw   = np.concatenate([X_nor_va, X_atk_va], axis=0)
X_test_raw  = np.concatenate([X_nor_te, X_atk_te], axis=0)
y_bin_train = np.concatenate([y_nor_bin_tr, y_atk_bin_tr])
y_bin_val   = np.concatenate([y_nor_bin_va, y_atk_bin_va])
y_bin_test  = np.concatenate([y_nor_bin_te, y_atk_bin_te])
y_type_train= np.concatenate([y_nor_id_tr, y_atk_id_tr])
y_type_val  = np.concatenate([y_nor_id_va, y_atk_id_va])
y_type_test = np.concatenate([y_nor_id_te, y_atk_id_te])

del df_normal, df_attack_only; gc.collect()

# ── Scaler: fit on NORMAL-ONLY train rows ────────────────────────────────────
scaler = RobustScaler()
scaler.fit(X_nor_train_raw)

X_train_sc     = scaler.transform(X_train_raw).astype(np.float32)
X_val_sc       = scaler.transform(X_val_raw).astype(np.float32)
X_test_sc      = scaler.transform(X_test_raw).astype(np.float32)
X_nor_train_sc = scaler.transform(X_nor_train_raw).astype(np.float32)

del X_train_raw, X_val_raw, X_test_raw, X_nor_train_raw; gc.collect()

N_FEATURES = X_train_sc.shape[1]
n_nor_tr   = (y_bin_train == 0).sum()
n_atk_tr   = (y_bin_train == 1).sum()

print(f"\nCombined splits (normal D1 + attack D2):")
print(f"  Train  : {X_train_sc.shape[0]:,} rows  — normal {n_nor_tr:,} ({n_nor_tr/len(y_bin_train)*100:.1f}%)"
      f"  attack {n_atk_tr:,} ({n_atk_tr/len(y_bin_train)*100:.1f}%)")
print(f"  Val    : {X_val_sc.shape[0]:,}  — normal {(y_bin_val==0).sum():,}  attack {(y_bin_val==1).sum():,}")
print(f"  Test   : {X_test_sc.shape[0]:,}  — normal {(y_bin_test==0).sum():,}  attack {(y_bin_test==1).sum():,}")
print(f"  N_FEATURES   : {N_FEATURES}")
print(f"  n_classes    : {n_classes}")
print(f"  Scaler fit on: {X_nor_train_sc.shape[0]:,} pure normal rows (Dataset 1 train split)")

# Sanity check
SPW_check = n_nor_tr / max(n_atk_tr, 1)
print(f"\nPRE-TRAINING SANITY CHECK")
print(f"  Normal fraction   : {n_nor_tr/len(y_bin_train)*100:.1f}%  (was 13.4% in v2 — Dataset 2 only)")
print(f"  scale_pos_weight  : {SPW_check:.3f}")
print(f"  CNN pos_weight    : {min(SPW_check, 4.0):.3f}  (capped 4.0)")
print("[OK] Ready to train")

# ── v4 FIX 1: Record normal→attack boundary indices for CNN window exclusion ──
# Without this, SwatWindowDataset creates windows that CROSS the boundary
# between the two concatenated time series, producing a trivially-detectable
# sensor discontinuity artifact that gives val_auc=1.0 from epoch 1.
NORMAL_TRAIN_BOUNDARY = len(X_nor_tr)   # index in X_train_sc where attack starts
NORMAL_VAL_BOUNDARY   = len(X_nor_va)   # index in X_val_sc where attack starts
NORMAL_TEST_BOUNDARY  = len(X_nor_te)   # index in X_test_sc where attack starts
print(f"[v4] Boundary indices — train: {NORMAL_TRAIN_BOUNDARY:,} | val: {NORMAL_VAL_BOUNDARY:,} | test: {NORMAL_TEST_BOUNDARY:,}")

# ── v5 FIX: Stratified per-attack-type undersampling with MIN_ROWS floor ──────
# v4 used uniform rng.choice which:
#   a) Created temporal gaps (false seams) the CNN learns as anomaly signals
#   b) Eliminated rare attack types (replay: ~207 rows → ~1 window)
# Fix: Undersample per attack type proportionally, with a minimum floor.
# Also track the gap indices created by undersampling for boundary exclusion.
import pandas as pd
TARGET_ATTACK_FRAC = 0.25
n_nor_tr_count = len(X_nor_tr)
n_atk_target   = int(n_nor_tr_count * TARGET_ATTACK_FRAC / (1 - TARGET_ATTACK_FRAC))
ATK_GAP_BOUNDARIES = []

if len(X_atk_tr) > n_atk_target:
    rng_sub = np.random.default_rng(SEED)
    MIN_ROWS_PER_TYPE = max(SEQ_LEN * 10, 1500)  # at least 10 non-overlapping windows

    df_atk_idx = pd.DataFrame({"y_id": y_atk_id_tr})
    groups = []
    for aid, grp in df_atk_idx.groupby("y_id"):
        target = max(int(n_atk_target * len(grp) / len(df_atk_idx)), MIN_ROWS_PER_TYPE)
        target = min(target, len(grp))
        sampled_idx = np.sort(rng_sub.choice(grp.index.values, size=target, replace=False))
        groups.append(sampled_idx)

    atk_sub_idx    = np.sort(np.concatenate(groups))
    X_atk_tr       = X_atk_tr[atk_sub_idx]
    y_atk_bin_tr   = y_atk_bin_tr[atk_sub_idx]
    y_atk_id_tr    = y_atk_id_tr[atk_sub_idx]
    print(f"[v5] Attack train stratified-undersampled to {len(atk_sub_idx):,} rows")
    print(f"     (MIN_ROWS_PER_TYPE={MIN_ROWS_PER_TYPE})")

    # Detect all temporal gap positions within the subsampled attack data
    atk_gaps = np.where(np.diff(atk_sub_idx) > 1)[0] + 1
    ATK_GAP_BOUNDARIES = list(atk_gaps + n_nor_tr_count)
    print(f"     Detected {len(ATK_GAP_BOUNDARIES)} temporal gaps in subsampled attack data")

NORMAL_TRAIN_BOUNDARY = n_nor_tr_count

````

## Layer 1a — Isolation Forest

**v3 changes:**
- `IF_FPR_TARGET = 0.08` (was 0.03 — 8% FPR gives better recall)
- IF score added as extra feature column for XGBoost input (`X_train_aug`, etc.)
- IF is NOT a voting layer — it contributes its anomaly score as a feature


````python
#cell 6
print("Training Isolation Forest on normal-only data...")
t0 = time.time()
iso_forest = IsolationForest(
    n_estimators=200, contamination=0.05,
    max_samples="auto", random_state=SEED, n_jobs=-1
)
iso_forest.fit(X_nor_train_sc)
print(f"IF trained in {time.time()-t0:.1f}s on {X_nor_train_sc.shape[0]:,} samples")


def if_score(model, X):
    """Normalise IF decision_function to [0,1]: high = anomalous."""
    raw = model.decision_function(X)
    return (1.0 - (raw - raw.min()) / (raw.max() - raw.min() + 1e-9)).astype(np.float32)


iso_train = if_score(iso_forest, X_train_sc)
iso_val   = if_score(iso_forest, X_val_sc)
iso_test  = if_score(iso_forest, X_test_sc)

# Calibrate: FPR < IF_FPR_TARGET on val normal rows (8%, was 3%)
val_nor_sc = iso_val[y_bin_val == 0]
thresholds = np.linspace(0.40, 0.97, 400)
fprs_t     = [(val_nor_sc > t).mean() for t in thresholds]
valid_t    = [t for t, f in zip(thresholds, fprs_t) if f < IF_FPR_TARGET]
IF_THRESH  = float(min(valid_t)) if valid_t else 0.55
l1_val     = (iso_val  > IF_THRESH).astype(np.int8)
l1_test    = (iso_test > IF_THRESH).astype(np.int8)
fpr_nor    = (val_nor_sc > IF_THRESH).mean()
atk_recall = l1_val[y_bin_val == 1].mean() if (y_bin_val==1).sum() > 0 else 0.0

print(f"\nIF threshold   : {IF_THRESH:.3f}  (FPR target < {IF_FPR_TARGET*100:.0f}%)")
print(f"FPR (nor val)  : {fpr_nor*100:.2f}%")
print(f"Attack recall  : {atk_recall*100:.1f}%  (IF alone — used as feature not voter)")

# ── v3 FIX: Augment feature matrices with IF score (RC-IF-4 fix) ──────────────
# IF score becomes an extra feature column so XGBoost can learn to use it
# This converts a broken voting layer into a useful anomaly-signal feature
iso_score_tr   = iso_train.reshape(-1, 1).astype(np.float32)
iso_score_val  = iso_val.reshape(-1, 1).astype(np.float32)
iso_score_test = iso_test.reshape(-1, 1).astype(np.float32)

X_train_aug = np.hstack([X_train_sc, iso_score_tr])
X_val_aug   = np.hstack([X_val_sc,   iso_score_val])
X_test_aug  = np.hstack([X_test_sc,  iso_score_test])

print(f"\nAugmented feature shape: {X_train_aug.shape}  (+1 IF score column)")

fig, axes = plt.subplots(1, 2, figsize=(14, 4))
axes[0].hist(val_nor_sc, bins=80, alpha=0.6, color="#2ecc71", label="Normal", density=True)
if (y_bin_val==1).sum() > 0:
    axes[0].hist(iso_val[y_bin_val==1], bins=80, alpha=0.6, color="#e74c3c", label="Attack", density=True)
axes[0].axvline(IF_THRESH, color="black", lw=1.5, linestyle="--", label=f"Thresh={IF_THRESH:.3f}")
axes[0].set_title("IF Score Distribution (val)"); axes[0].legend()
axes[1].plot(thresholds, [f*100 for f in fprs_t], lw=2)
axes[1].axhline(IF_FPR_TARGET*100, color="red", linestyle="--", lw=1.5, label=f"{IF_FPR_TARGET*100:.0f}% target")
axes[1].axvline(IF_THRESH, color="black", linestyle="--", lw=1.5)
axes[1].set_title("FPR vs Threshold"); axes[1].set_xlabel("Threshold"); axes[1].legend()
plt.suptitle("Layer 1a — Isolation Forest", fontsize=12, fontweight="bold")
plt.tight_layout(); plt.show()
````

## Layer 1b — Autoencoder (NEW — was missing in v2)

**Why:** Research report (Section 1.5) specifies Autoencoder as a second anomaly detector
trained on normal-only data (128→64→32→64→128). The reconstruction error over normal
data defines a second anomaly score, complementary to IF.

**Architecture:**
```
Input (N_FEATURES+1) → Dense(128, ReLU) → Dropout(0.10)
→ Dense(64, ReLU) → Dense(32, ReLU)          ← bottleneck
→ Dense(64, ReLU) → Dense(128, ReLU)
→ Dense(N_FEATURES+1, linear)                 ← reconstruction
Loss: MSE on normal-only data
Anomaly score: reconstruction error on new sample
```

**Key property:** Normal samples have low reconstruction error (AE learned their manifold).
Attack samples have high reconstruction error (AE can't reconstruct out-of-distribution data).
The AE score is added as a second augmented feature alongside IF score for XGBoost.


````python
#cell 7
# =============================================================================
# PYTORCH AUTOENCODER — trained on normal-only data
# Architecture: 128 → 64 → 32 → 64 → 128 (from research report Section 1.5)
# =============================================================================

import torch
import torch.nn as nn
import numpy as np

class SWaTAutoencoder(nn.Module):
    def __init__(self, input_dim):
        super(SWaTAutoencoder, self).__init__()
        # Ensure the input_dim matches your X_nor_aug.shape[1]
        self.encoder = nn.Sequential(
            nn.Linear(input_dim, 128),
            nn.ReLU(),
            nn.Linear(128, 64),
            nn.ReLU(),
            nn.Linear(64, 32) # Latent space
        )
        self.decoder = nn.Sequential(
            nn.Linear(32, 64),
            nn.ReLU(),
            nn.Linear(64, 128),
            nn.ReLU(),
            nn.Linear(128, input_dim)
        )

    def forward(self, x):
        z = self.encoder(x)
        x_r = self.decoder(z)
        return x_r

    def reconstruction_error(self, x):
        """Per-sample MSE reconstruction error (anomaly score)."""
        with torch.no_grad():
            x_r = self.forward(x)
            return ((x - x_r) ** 2).mean(dim=1)


# ── Dataset for AE — simple flat tensors ────────────────────────────────────
class FlatDataset(Dataset):
    def __init__(self, X): self.X = torch.from_numpy(X).float()
    def __len__(self): return len(self.X)
    def __getitem__(self, i): return self.X[i]


ae_model  = SWaTAutoencoder(X_train_aug.shape[1]).to(DEVICE)
ae_params = sum(p.numel() for p in ae_model.parameters())
print(f"SWaTAutoencoder: {ae_params:,} params | input_dim={X_train_aug.shape[1]}")

# Train on normal-only data
ae_nor_ds     = FlatDataset(X_nor_train_sc[:, :X_train_aug.shape[1] - 1]
                             if X_nor_train_sc.shape[1] < X_train_aug.shape[1]
                             else X_nor_train_sc)
# Augment normal training data with IF score too
nor_if_score  = if_score(iso_forest, X_nor_train_sc).reshape(-1, 1)
X_nor_aug     = np.hstack([X_nor_train_sc, nor_if_score])
ae_nor_ds     = FlatDataset(X_nor_aug)
ae_loader     = DataLoader(ae_nor_ds, batch_size=256, shuffle=True, num_workers=0)

ae_optimizer  = torch.optim.Adam(ae_model.parameters(), lr=1e-3, weight_decay=1e-5)
ae_scheduler  = torch.optim.lr_scheduler.ReduceLROnPlateau(
    ae_optimizer, mode="min", factor=0.5, patience=5, min_lr=1e-6)

# Early stopping on val reconstruction error (computed on normal val rows)
X_val_nor_aug  = X_val_aug[y_bin_val == 0]
val_nor_tensor = torch.from_numpy(X_val_nor_aug).float().to(DEVICE)

print("\nTraining Autoencoder on normal-only data...")
AE_EPOCHS  = 60
best_ae_loss, ae_patience_cnt, ae_patience = float("inf"), 0, 10
AE_CKPT    = str(MODELS_DIR / "layer1b_autoencoder_best.pt")
ae_losses  = []

for ep in range(1, AE_EPOCHS + 1):
    ae_model.train()
    ep_loss, ep_n = 0.0, 0
    for X_b in ae_loader:
        X_b = X_b.to(DEVICE)
        ae_optimizer.zero_grad()
        X_r  = ae_model(X_b)
        loss = F.mse_loss(X_r, X_b)
        loss.backward()
        torch.nn.utils.clip_grad_norm_(ae_model.parameters(), 1.0)
        ae_optimizer.step()
        ep_loss += loss.item() * len(X_b)
        ep_n    += len(X_b)
    train_mse = ep_loss / ep_n

    # Validation: reconstruction error on normal val rows
    ae_model.eval()
    with torch.no_grad():
        val_r   = ae_model(val_nor_tensor)
        val_mse = F.mse_loss(val_r, val_nor_tensor).item()
    ae_losses.append(train_mse)
    ae_scheduler.step(val_mse)

    if val_mse < best_ae_loss:
        best_ae_loss = val_mse
        torch.save(ae_model.state_dict(), AE_CKPT)
        ae_patience_cnt = 0
    else:
        ae_patience_cnt += 1
        if ae_patience_cnt >= ae_patience:
            print(f"  AE early stop at epoch {ep}  (best val_mse={best_ae_loss:.6f})")
            break

    if ep % 10 == 0 or ep == 1:
        print(f"  AE Ep {ep:3d} | train_mse={train_mse:.6f} | val_mse={val_mse:.6f}")

ae_model.load_state_dict(torch.load(AE_CKPT, map_location=DEVICE))
print(f"\nAutoencoder trained | best val_mse={best_ae_loss:.6f}")

# ── Compute AE reconstruction error as anomaly score ────────────────────────
def ae_anomaly_score(model, X_np, device, batch_size=512):
    """Compute per-sample reconstruction error (anomaly score)."""
    model.eval()
    scores = []
    for i in range(0, len(X_np), batch_size):
        xb = torch.from_numpy(X_np[i:i+batch_size]).float().to(device)
        with torch.no_grad():
            err = model.reconstruction_error(xb)
        scores.append(err.cpu().numpy())
    return np.concatenate(scores).astype(np.float32)


ae_score_tr   = ae_anomaly_score(ae_model, X_train_aug, DEVICE)
ae_score_val  = ae_anomaly_score(ae_model, X_val_aug,   DEVICE)
ae_score_test = ae_anomaly_score(ae_model, X_test_aug,  DEVICE)

# Normalise to [0,1]
ae_min, ae_max = ae_score_tr.min(), ae_score_tr.max()
ae_score_tr_n  = ((ae_score_tr  - ae_min) / (ae_max - ae_min + 1e-9)).clip(0, 1)
ae_score_val_n = ((ae_score_val - ae_min) / (ae_max - ae_min + 1e-9)).clip(0, 1)
ae_score_test_n= ((ae_score_test- ae_min) / (ae_max - ae_min + 1e-9)).clip(0, 1)

# Calibrate AE threshold on val (FPR < 5%)
ae_nor_scores = ae_score_val_n[y_bin_val == 0]
ae_thrs       = np.linspace(0.05, 0.99, 400)
ae_fprs       = [(ae_nor_scores > t).mean() for t in ae_thrs]
ae_valid      = [t for t, f in zip(ae_thrs, ae_fprs) if f < 0.05]
AE_THRESHOLD  = float(min(ae_valid)) if ae_valid else 0.50
ae_recall_atk = ae_score_val_n[y_bin_val==1].mean() if (y_bin_val==1).sum() > 0 else 0.0

print(f"\nAE threshold   : {AE_THRESHOLD:.3f}  (FPR target < 5%)")
print(f"AE FPR (nor)   : {(ae_nor_scores > AE_THRESHOLD).mean()*100:.2f}%")
print(f"AE attack score: mean={ae_score_val_n[y_bin_val==1].mean():.3f}  "
      f"(normal mean={ae_nor_scores.mean():.3f})")

# Add AE score as additional feature column (second anomaly score alongside IF)
X_train_aug = np.hstack([X_train_aug, ae_score_tr_n.reshape(-1, 1)])
X_val_aug   = np.hstack([X_val_aug,   ae_score_val_n.reshape(-1, 1)])
X_test_aug  = np.hstack([X_test_aug,  ae_score_test_n.reshape(-1, 1)])
print(f"\nDouble-augmented feature shape: {X_train_aug.shape}  (+IF score +AE score)")

# Plot
fig, axes = plt.subplots(1, 2, figsize=(14, 4))
axes[0].hist(ae_nor_scores, bins=80, alpha=0.6, color="#2ecc71", label="Normal", density=True)
if (y_bin_val==1).sum() > 0:
    axes[0].hist(ae_score_val_n[y_bin_val==1], bins=80, alpha=0.6, color="#e74c3c", label="Attack", density=True)
axes[0].axvline(AE_THRESHOLD, color="black", lw=1.5, linestyle="--", label=f"Thresh={AE_THRESHOLD:.3f}")
axes[0].set_title("AE Reconstruction Error (val)"); axes[0].legend()
axes[1].plot(ae_losses, lw=2)
axes[1].set_title("AE Training Loss (MSE)"); axes[1].set_xlabel("Epoch")
plt.suptitle("Layer 1b — Autoencoder", fontsize=12, fontweight="bold")
plt.tight_layout(); plt.show()

# ── v4 FIX 2: Compute AE reconstruction error on NORMAL-ONLY val data ─────────
# Also use percentile-based threshold instead of FPR sweep (which gave threshold
# 50× higher than any actual score, making AE catch nothing)

def ae_recon_errors_batch(ae_model, X_np, device, batch_size=512):
    """Compute per-sample MSE reconstruction error (raw, un-normalized)."""
    ae_model.eval()
    errors = []
    with torch.no_grad():
        for i in range(0, len(X_np), batch_size):
            xb = torch.from_numpy(X_np[i:i+batch_size]).to(device)
            recon = ae_model(xb)
            mse   = ((recon - xb) ** 2).mean(dim=1).cpu().numpy()
            errors.append(mse)
    return np.concatenate(errors)

# Normal-only validation data (scaled)
X_nor_va_sc_pure = scaler.transform(X_nor_va).astype(np.float32)

# Compute errors on normal train (for threshold calibration)
nor_train_errors = ae_recon_errors_batch(ae_model, X_nor_aug, DEVICE)
# Compute errors on attack train sample (for sanity check)
atk_sample_sc    = scaler.transform(X_atk_tr[:min(5000, len(X_atk_tr))]).astype(np.float32)
atk_train_errors = ae_recon_errors_batch(ae_model, atk_sample_sc, DEVICE)

# v4 FIX: percentile-based threshold (99th pct of NORMAL errors)
AE_THRESHOLD_RAW = float(np.percentile(nor_train_errors, 99))
print(f"\n[v4] AE threshold (99th pct of normal train): {AE_THRESHOLD_RAW:.6f}")
print(f"[v4] Normal train error  — mean={nor_train_errors.mean():.6f}  p99={AE_THRESHOLD_RAW:.6f}")
print(f"[v4] Attack sample error — mean={atk_train_errors.mean():.6f}")

if atk_train_errors.mean() < nor_train_errors.mean() * 1.5:
    print("[WARN] AE attack errors NOT significantly higher than normal.")
    print("       AE is reconstructing attacks too well — check architecture or training data.")
else:
    print("[OK]  AE separates normal vs attack reconstruction errors.")

# Rebuild normalized scores using the correct threshold
# Normalize: score = raw_error / AE_THRESHOLD_RAW  (>1.0 means anomalous)
def ae_score_normalized(ae_model, X_np, threshold_raw, device, batch_size=512):
    raw = ae_recon_errors_batch(ae_model, X_np, device, batch_size)
    return np.clip(raw / (threshold_raw + 1e-9), 0.0, 1.0).astype(np.float32)

# Recompute AE scores on all splits with corrected normalization
ae_score_train_n = ae_score_normalized(ae_model, X_train_aug, AE_THRESHOLD_RAW, DEVICE)
ae_score_val_n   = ae_score_normalized(ae_model, X_val_aug,   AE_THRESHOLD_RAW, DEVICE)
ae_score_test_n  = ae_score_normalized(ae_model, X_test_aug,  AE_THRESHOLD_RAW, DEVICE)

# Threshold for binary AE prediction is now simply 1.0 (raw exceeds threshold)
AE_THRESHOLD = 1.0
ae_fpr_val = ((ae_score_val_n[:NORMAL_VAL_BOUNDARY] > 1.0).sum() /
               max(NORMAL_VAL_BOUNDARY, 1))
ae_recall_val = ((ae_score_val_n[NORMAL_VAL_BOUNDARY:] > 1.0).sum() /
                  max(len(ae_score_val_n) - NORMAL_VAL_BOUNDARY, 1))
print(f"[v4] AE FPR on val normal:  {ae_fpr_val*100:.2f}%")
print(f"[v4] AE recall on val attack: {ae_recall_val*100:.2f}%")
````

````python
#cell 9
print(f"Training XGBoost ({XGB_DEVICE.upper()} | hist | binary:logistic | scale_pos_weight)...")
t0 = time.time()

xgb_params = dict(
    objective         = "binary:logistic",
    eval_metric       = "aucpr",          # AUC-PR: robust under imbalance
    tree_method       = XGB_TREE_METHOD,
    device            = XGB_DEVICE,
    n_jobs            = XGB_N_JOBS,
    n_estimators      = 800,              # was 500 — stopped at 72, needs more
    max_depth         = 5,               # was 7 — shallower prevents overfitting
    learning_rate     = 0.03,            # was 0.05 — slower, more detail
    subsample         = 0.8,
    colsample_bytree  = 0.7,             # was 0.8
    min_child_weight  = 10,              # was 5 — larger min leaf
    reg_alpha         = 0.5,             # was 0.1
    reg_lambda        = 2.0,             # was 1.0
    scale_pos_weight  = SPW,             # handles imbalance correctly
    early_stopping_rounds = 50,          # was 30 — more patience
    random_state      = SEED,
)

xgb_model = xgb.XGBClassifier(**xgb_params)
xgb_model.fit(
    X_xgb_train, y_xgb_train,
    eval_set=[(X_xgb_val, y_xgb_val)],
    verbose=100,
)
print(f"XGBoost trained in {time.time()-t0:.1f}s | best_iter={xgb_model.best_iteration}")

xgb_prob_val  = xgb_model.predict_proba(X_xgb_val)[:, 1].astype(np.float32)
xgb_prob_test = xgb_model.predict_proba(X_xgb_test)[:, 1].astype(np.float32)

# v3 FIX: calibrate threshold on val F1 (was hardcoded 0.70)
thresholds_xgb = np.linspace(0.20, 0.85, 500)
f1s_xgb = [f1_score(y_bin_val, (xgb_prob_val > t).astype(int), zero_division=0)
            for t in thresholds_xgb]
XGB_THRESHOLD = float(thresholds_xgb[np.argmax(f1s_xgb)])
xgb_pred_val  = (xgb_prob_val  > XGB_THRESHOLD).astype(np.int8)
xgb_pred_test = (xgb_prob_test > XGB_THRESHOLD).astype(np.int8)

f1_xgb  = f1_score(y_bin_val, xgb_pred_val)
auc_xgb = roc_auc_score(y_bin_val, xgb_prob_val)
print(f"\nXGBoost Val — F1: {f1_xgb:.4f} | AUC-PR: {auc_xgb:.4f}")
print(f"Calibrated threshold: {XGB_THRESHOLD:.3f}  (was hardcoded 0.70)")
print(classification_report(y_bin_val, xgb_pred_val, target_names=["Normal","Attack"]))

fig, axes = plt.subplots(1, 2, figsize=(14, 4))
axes[0].plot(thresholds_xgb, f1s_xgb, lw=2)
axes[0].axvline(XGB_THRESHOLD, color="red", linestyle="--", lw=1.5, label=f"Optimal={XGB_THRESHOLD:.3f}")
axes[0].set_title("XGBoost: Val F1 vs Threshold"); axes[0].set_xlabel("Threshold"); axes[0].legend()
fpr_r, tpr_r, _ = roc_curve(y_bin_val, xgb_prob_val)
axes[1].plot(fpr_r, tpr_r, lw=2)
axes[1].plot([0,1],[0,1],"k--",lw=0.8)
axes[1].set_title(f"XGBoost ROC (AUC={auc_xgb:.4f})"); axes[1].set_xlabel("FPR"); axes[1].set_ylabel("TPR")
plt.suptitle("Layer 2 — XGBoost", fontsize=12, fontweight="bold")
plt.tight_layout(); plt.show()
````

## Layer 3 — PyTorch CNN-BiLSTM + Attention

**Architecture (preserved from v2):**
```
Input: (batch, 150, N_FEATURES)
Conv1D(64, k=3, causal) → BN → ReLU
Conv1D(128, k=3, causal) → BN → Dropout(0.15)
MaxPool1D(2)
BiLSTM(128) → BN → Dropout(0.20)
BiLSTM(64) → Dropout(0.15)
TemporalAttention(128)
Dense(64, ReLU) → Dropout(0.15) → Dense(1, Sigmoid)
```

**v3 training fixes (from failure analysis Part 3):**
- Adam LR: `1e-3 → 3e-4` (RC-CNN-3: LR too high caused val_auc oscillation)
- EarlyStopping: `patience=8 → 15`, monitors `val_auc` not `val_loss` (RC-CNN-4)
- ReduceLROnPlateau: monitors `val_auc`, `mode=max`, `patience=5` (was val_loss)
- Epochs: `50 → 80` (EarlyStopping controls actual stopping)
- CNN uses `X_train_sc` (original features, not augmented) — temporal patterns only


````python
#cell 11
class SwatWindowDataset(Dataset):
    """Sliding window dataset. Majority vote labeling (not any()).

    Majority vote (>50% of timesteps attack) is used instead of any() because:
    - any() labels boundary windows (1-2 attack steps in 150) as attack
    - These windows still look normal — creates noisy labels
    - Majority vote: only windows truly dominated by attack get label=1

    v4 FIX: boundary_indices parameter excludes windows that cross dataset
    boundaries (e.g. the seam between normal_train and attack_train arrays).
    Without this, the CNN detects the sensor discontinuity artifact rather than
    genuine attack signatures, producing val_auc=1.0 from epoch 1.
    """
    def __init__(self, X, y_bin, seq_len, indices=None, majority_thresh=0.5,
                 boundary_indices=None):
        self.X = X; self.y = y_bin; self.seq_len = seq_len
        self.maj_thresh = majority_thresh
        n_windows = len(X) - seq_len

        if indices is not None:
            starts = np.array([i for i in indices if i < n_windows], dtype=np.int64)
        else:
            starts = np.arange(n_windows, dtype=np.int64)

        # v4 FIX: Exclude windows that cross any dataset boundary
        if boundary_indices is not None:
            boundary_set = set(boundary_indices)
            def crosses_boundary(s):
                end = s + seq_len
                return any(s < b <= end for b in boundary_set)
            n_before = len(starts)
            starts = np.array([s for s in starts if not crosses_boundary(s)], dtype=np.int64)
            n_excluded = n_before - len(starts)
            if n_excluded > 0:
                print(f"  [v4] Excluded {n_excluded:,} boundary-crossing windows "
                      f"({n_excluded/n_before*100:.1f}% of all windows)")

        self.starts = starts

    def __len__(self): return len(self.starts)

    def __getitem__(self, idx):
        s   = self.starts[idx]
        X_w = torch.from_numpy(self.X[s : s+self.seq_len])
        frac = float(self.y[s : s+self.seq_len].mean())
        y_w  = torch.tensor(1.0 if frac > self.maj_thresh else 0.0)
        return X_w, y_w

    def get_labels(self):
        return np.array(
            [1 if self.y[s:s+self.seq_len].mean() > self.maj_thresh else 0
             for s in self.starts], dtype=np.int8
        )


# v5 FIX: Pass ALL boundary indices (normal-attack seam + undersampling gaps)
ALL_TRAIN_BOUNDARIES = [NORMAL_TRAIN_BOUNDARY] + ATK_GAP_BOUNDARIES
print(f"[v5] Total boundary indices for train: {len(ALL_TRAIN_BOUNDARIES)}")
train_ds = SwatWindowDataset(X_train_sc, y_bin_train, SEQ_LEN, majority_thresh=0.5,
                              boundary_indices=ALL_TRAIN_BOUNDARIES)
val_ds   = SwatWindowDataset(X_val_sc,   y_bin_val,   SEQ_LEN, majority_thresh=0.5,
                              boundary_indices=[NORMAL_VAL_BOUNDARY])
test_ds  = SwatWindowDataset(X_test_sc,  y_bin_test,  SEQ_LEN, majority_thresh=0.5,
                              boundary_indices=[NORMAL_TEST_BOUNDARY])

train_labels = train_ds.get_labels()
n_train_nor  = (train_labels == 0).sum()
n_train_atk  = (train_labels == 1).sum()

# v4 FIX: Subsample training windows to MAX_TRAIN_WINDOWS for CPU feasibility
MAX_TRAIN_WINDOWS = MAX_TRAIN_WINDOWS  # from config cell — 1M on A6000, 300K on CPU
if len(train_ds) > MAX_TRAIN_WINDOWS:
    rng_win = np.random.default_rng(SEED)
    sampled_starts = np.sort(rng_win.choice(train_ds.starts, size=MAX_TRAIN_WINDOWS, replace=False))
    train_ds.starts = sampled_starts
    train_labels    = train_ds.get_labels()
    n_train_nor     = (train_labels == 0).sum()
    n_train_atk     = (train_labels == 1).sum()
    print(f"[v4] CNN train windows subsampled to {MAX_TRAIN_WINDOWS:,} "
          f"(normal: {n_train_nor:,} | attack: {n_train_atk:,})")

sample_weights = np.where(train_labels==1, n_train_nor/max(n_train_atk,1), 1.0)
sampler = WeightedRandomSampler(
    weights=torch.from_numpy(sample_weights).float(),
    num_samples=len(train_labels), replacement=True
)

train_loader = DataLoader(train_ds, batch_size=BATCH_SIZE, sampler=sampler, num_workers=DATALOADER_WORKERS,
                          pin_memory=(DEVICE.type=="cuda"))
val_loader   = DataLoader(val_ds,   batch_size=BATCH_SIZE, shuffle=False, num_workers=DATALOADER_WORKERS,
                          pin_memory=(DEVICE.type=="cuda"))
test_loader  = DataLoader(test_ds,  batch_size=BATCH_SIZE, shuffle=False, num_workers=DATALOADER_WORKERS,
                          pin_memory=(DEVICE.type=="cuda"))

seq_labels_val  = val_ds.get_labels()
seq_labels_test = test_ds.get_labels()

print(f"Dataset: train {len(train_ds):,} | val {len(val_ds):,} | test {len(test_ds):,}")
print(f"Train windows: normal {n_train_nor:,} | attack {n_train_atk:,}")
print(f"Sampler weight for attack: {n_train_nor/max(n_train_atk,1):.2f}x")

# v4 SANITY CHECK: val_auc at epoch 1 should not be 1.0
# (This check prints in the training loop — added below)
print("\n[v4 SANITY] If val_auc=1.0000 at epoch 1 → boundary leakage still present.")
print("            Expected realistic epoch-1 val_auc: 0.55–0.80")
````

````python
#cell 12
class TemporalAttention(nn.Module):
    def __init__(self, feat_dim):
        super().__init__()
        self.W = nn.Linear(feat_dim, 1, bias=False)

    def forward(self, x):
        e = torch.tanh(self.W(x))        # (batch, time, 1)
        a = torch.softmax(e, dim=1)      # (batch, time, 1)
        return (x * a).sum(dim=1)        # (batch, feat)


class SWaTCNNBiLSTM(nn.Module):
    def __init__(self, n_features, seq_len):
        super().__init__()
        self.conv1    = nn.Conv1d(n_features, 64,  kernel_size=3, padding=2)
        self.bn1      = nn.BatchNorm1d(64)
        self.conv2    = nn.Conv1d(64, 128, kernel_size=3, padding=2)
        self.bn2      = nn.BatchNorm1d(128)
        self.drop_cnn = nn.Dropout(0.15)
        self.pool     = nn.MaxPool1d(2)
        self.bilstm1  = nn.LSTM(128, 128, batch_first=True, bidirectional=True)
        self.bn3      = nn.BatchNorm1d(256)
        self.drop1    = nn.Dropout(0.20)   # was 0.30 (RC-CNN-1)
        self.bilstm2  = nn.LSTM(256, 64,  batch_first=True, bidirectional=True)
        self.drop2    = nn.Dropout(0.15)   # was 0.25
        self.attention = TemporalAttention(128)
        self.fc1      = nn.Linear(128, 64)
        self.drop3    = nn.Dropout(0.15)
        self.fc2      = nn.Linear(64, 1)
        self._init_weights()

    def _init_weights(self):
        for m in self.modules():
            if isinstance(m, nn.Conv1d):   nn.init.kaiming_normal_(m.weight, nonlinearity="relu")
            elif isinstance(m, nn.Linear): nn.init.xavier_uniform_(m.weight)
            elif isinstance(m, nn.LSTM):
                for name, p in m.named_parameters():
                    if "weight" in name: nn.init.orthogonal_(p)
                    elif "bias"  in name: nn.init.zeros_(p)

    def forward(self, x):
        # x: (batch, seq_len, n_features)
        h = x.permute(0, 2, 1)                    # (batch, feat, time)
        h = F.relu(self.bn1(self.conv1(h)))
        h = h[:, :, :x.size(1)]                   # causal trim
        h = self.drop_cnn(F.relu(self.bn2(self.conv2(h))))
        h = h[:, :, :x.size(1)]                   # causal trim
        h = self.pool(h)                           # (batch, 128, time//2)
        h = h.permute(0, 2, 1)                    # (batch, time//2, 128)
        h, _ = self.bilstm1(h)                    # (batch, time//2, 256)
        h = self.bn3(h.permute(0,2,1)).permute(0,2,1)
        h = self.drop1(h)
        h, _ = self.bilstm2(h)                    # (batch, time//2, 128)
        h = self.drop2(h)
        h = self.attention(h)                      # (batch, 128)
        h = self.drop3(F.relu(self.fc1(h)))
        return torch.sigmoid(self.fc2(h))          # (batch, 1)


model  = SWaTCNNBiLSTM(N_FEATURES, SEQ_LEN).to(DEVICE)
params = sum(p.numel() for p in model.parameters())
print(f"SWaTCNNBiLSTM: {params:,} params = {params*4/1024**2:.1f} MB | device={DEVICE}")

with torch.no_grad():
    _x = torch.randn(2, SEQ_LEN, N_FEATURES).to(DEVICE)
    _y = model(_x)
    print(f"Forward pass: {tuple(_x.shape)} → {tuple(_y.shape)} ✅")
````

---
## Layer 3b — Training (v3: fixed LR, patience, correct monitor metric)

**v3 training fixes:**
- `Adam lr=3e-4` (was `1e-3` — RC-CNN-3: too high, val_auc oscillated 0.70→0.45→0.64)
- `EarlyStopping patience=15` (was 8 — model stopped at epoch 9, never converged)
- `ReduceLROnPlateau monitor=val_auc, mode=max, patience=5` (was val_loss — RC-CNN-4)
- `MAX_EPOCHS=80` (was 50)
- Loss: Weighted BCE with `pos_weight` (maintained from v2 — correct choice vs focal)


````python
#cell 13
# Weighted BCE loss — pos_weight inversely proportional to attack fraction
# Cap at 4.0 to prevent gradient instability
pos_weight_val = float(n_train_nor) / max(float(n_train_atk), 1.0)
pos_weight_val = min(pos_weight_val, 4.0)
pos_weight_t   = torch.tensor([pos_weight_val]).to(DEVICE)


def weighted_bce(pred, target, pos_weight):
    pred   = pred.squeeze()
    target = target.float()
    bce    = -(target * torch.log(pred.clamp(1e-7, 1-1e-7)) +
               (1-target) * torch.log((1-pred).clamp(1e-7, 1-1e-7)))
    weight = torch.where(target == 1, pos_weight, torch.ones_like(target))
    return (bce * weight).mean()


# v3 FIX: lr=3e-4 (was 1e-3), clipnorm=1.0
optimizer = torch.optim.Adam(model.parameters(), lr=3e-4, weight_decay=1e-5)

# A6000: Automatic Mixed Precision (AMP) — halves memory, ~1.5–2× faster on Ampere
# Uses BF16 for forward pass, FP32 for gradient accumulation
amp_scaler = torch.cuda.amp.GradScaler(enabled=USE_AMP)
if USE_AMP:
    print(f"[AMP] GradScaler enabled — BF16 forward pass on {DEVICE}")
else:
    print(f"[AMP] Disabled (CPU or pre-Ampere GPU)")
# v3 FIX: monitor val_auc (mode=max), patience=5 (was val_loss, patience=3)
scheduler = torch.optim.lr_scheduler.ReduceLROnPlateau(
    optimizer, mode="max", factor=0.5, patience=5, min_lr=1e-6
)


def train_epoch(model, loader, optimizer, pos_weight, device):
    model.train()
    total_loss, total = 0.0, 0
    for X_b, y_b in loader:
        X_b, y_b = X_b.to(device), y_b.to(device)
        optimizer.zero_grad()
        # A6000: AMP context — BF16 forward, FP32 gradients
        with torch.cuda.amp.autocast(enabled=USE_AMP):
            pred = model(X_b).squeeze()
            loss = weighted_bce(pred, y_b, pos_weight)
        amp_scaler.scale(loss).backward()
        amp_scaler.unscale_(optimizer)
        torch.nn.utils.clip_grad_norm_(model.parameters(), max_norm=1.0)
        amp_scaler.step(optimizer)
        amp_scaler.update()
        total_loss += loss.item() * len(X_b)
        total      += len(X_b)
    return total_loss / total


def evaluate(model, loader, device):
    model.eval()
    all_preds, all_labels = [], []
    total_loss, total = 0.0, 0
    with torch.no_grad():
        for X_b, y_b in loader:
            X_b, y_b = X_b.to(device), y_b.to(device)
            with torch.cuda.amp.autocast(enabled=USE_AMP):
                pred = model(X_b).squeeze()
                loss = weighted_bce(pred, y_b, pos_weight_t.squeeze())
            total_loss += loss.item() * len(X_b)
            total      += len(X_b)
            all_preds.append(pred.cpu().numpy())
            all_labels.append(y_b.cpu().numpy())
    probs  = np.concatenate(all_preds)
    labels = np.concatenate(all_labels).astype(np.int8)
    auc    = roc_auc_score(labels, probs) if labels.sum() > 0 and (labels==0).sum() > 0 else 0.0
    return total_loss/total, auc, probs, labels


# ── Training loop ─────────────────────────────────────────────────────────────
CKPT      = str(MODELS_DIR / "layer3_cnn_lstm_best.pt")
best_auc  = 0.0
p_cnt     = 0
leakage_detected = False
PATIENCE  = 15    # was 8 (RC-CNN-4: stopped too early)
MAX_EPOCHS = 80   # was 50; EarlyStopping controls actual stop
train_losses, val_aucs = [], []

print(f"Training CNN-BiLSTM | device={DEVICE} | lr=3e-4 | patience={PATIENCE} | max_epochs={MAX_EPOCHS}")
print(f"pos_weight={pos_weight_val:.2f}  SEQ_LEN={SEQ_LEN}\n")

epoch_t0 = time.time()
for epoch in range(1, MAX_EPOCHS + 1):
    t_ep       = time.time()
    tr_loss    = train_epoch(model, train_loader, optimizer, pos_weight_t.squeeze(), DEVICE)
    vl_loss, vl_auc, _, _ = evaluate(model, val_loader, DEVICE)
    train_losses.append(tr_loss)
    val_aucs.append(vl_auc)
    elapsed = time.time() - t_ep
    print(f"Ep {epoch:3d}/{MAX_EPOCHS} | train_loss={tr_loss:.4f} | val_auc={vl_auc:.4f} | {elapsed:.0f}s")
    # v4 SANITY CHECK: Perfect val_auc from epoch 1 = data leakage
    if epoch == 1 and vl_auc >= 0.9999:
        leakage_detected = True
        print("  [CRITICAL WARNING] val_auc=1.0 at epoch 1 — DATA LEAKAGE DETECTED.")
        print("  Check that boundary_indices were passed to SwatWindowDataset.")
        print("  Training will continue but results will NOT be valid.")
    # v5 FIX: Only suppress scheduler if leakage was detected at epoch 1
    if not leakage_detected:
        scheduler.step(vl_auc)
    if vl_auc > best_auc:
        best_auc = vl_auc
        torch.save(model.state_dict(), CKPT)
        print(f"  ↑ Best model saved (val_auc={best_auc:.4f})")
        p_cnt = 0
    else:
        p_cnt += 1
        if p_cnt >= PATIENCE:
            print(f"\nEarly stop at epoch {epoch} (no val_auc improvement for {PATIENCE} epochs)")
            break

model.load_state_dict(torch.load(CKPT, map_location=DEVICE))
print(f"\nTraining done: {(time.time()-epoch_t0)/60:.1f} min | best val_auc={best_auc:.4f}")

fig, axes = plt.subplots(1, 2, figsize=(14, 4))
axes[0].plot(train_losses, lw=2, label="Train Loss")
axes[0].set_title("Weighted BCE Loss"); axes[0].set_xlabel("Epoch"); axes[0].legend()
axes[1].plot(val_aucs, lw=2, color="orange", label="Val AUC")
axes[1].axhline(best_auc, color="green", linestyle="--", lw=1.5, label=f"Best={best_auc:.4f}")
axes[1].set_title("Validation AUC"); axes[1].set_xlabel("Epoch"); axes[1].legend()
plt.suptitle("Layer 3 Training History", fontsize=12, fontweight="bold")
plt.tight_layout(); plt.show()
````

## Layer 3c — CNN-BiLSTM Evaluation + Threshold Calibration + Flagged-Window Inference


````python
#cell 14
# Calibrate CNN threshold on val F1 (was hardcoded 0.50 — RC-CNN-5)
_, _, val_cnn_probs, val_cnn_labels = evaluate(model, val_loader, DEVICE)

thresholds_cnn = np.linspace(0.05, 0.90, 400)
f1s_cnn = [f1_score(val_cnn_labels, (val_cnn_probs > t).astype(int), zero_division=0)
            for t in thresholds_cnn]
CNN_THRESHOLD = float(thresholds_cnn[np.argmax(f1s_cnn)])
print(f"Calibrated CNN threshold: {CNN_THRESHOLD:.3f}  (was hardcoded 0.50)")

# Full test evaluation
_, _, test_cnn_probs, test_cnn_labels = evaluate(model, test_loader, DEVICE)
test_cnn_preds = (test_cnn_probs > CNN_THRESHOLD).astype(np.int8)

cnn_f1  = f1_score(test_cnn_labels, test_cnn_preds)
cnn_auc = roc_auc_score(test_cnn_labels, test_cnn_probs) if test_cnn_labels.sum() > 0 else 0.0

print(f"\nCNN-BiLSTM Test:")
print(f"  F1: {cnn_f1:.4f} | AUC: {cnn_auc:.4f}")
print(classification_report(test_cnn_labels, test_cnn_preds, target_names=["Normal","Attack"]))

# ── Flagged-window inference (per original docx spec) ─────────────────────────
# CNN runs at TRAINING time on ALL windows (learns normal patterns)
# At INFERENCE: only windows flagged by IF OR XGB get CNN refinement
OFF         = SEQ_LEN
n_test_seqs = len(test_cnn_labels)

if_test_aligned  = l1_test[OFF:OFF+n_test_seqs].astype(np.float32)
xgb_test_aligned = xgb_prob_test[OFF:OFF+n_test_seqs].astype(np.float32)
y_type_test_seq  = y_type_test[OFF:OFF+n_test_seqs]

flagged_mask = (if_test_aligned > IF_THRESH) | (xgb_test_aligned > XGB_THRESHOLD)
flagged_idx  = np.where(flagged_mask)[0]
print(f"\nFlagged by IF or XGB: {flagged_mask.sum():,} / {n_test_seqs:,} windows ({flagged_mask.mean()*100:.1f}%)")

# Run CNN on flagged windows only
flagged_global = flagged_idx + OFF
flagged_ds     = SwatWindowDataset(X_test_sc, y_bin_test, SEQ_LEN, indices=flagged_global)
flagged_loader = DataLoader(flagged_ds, batch_size=BATCH_SIZE, shuffle=False, num_workers=0)
_, _, flagged_probs, flagged_labels = evaluate(model, flagged_loader, DEVICE)
print(f"CNN flagged-only: {len(flagged_probs):,} windows | attack={flagged_labels.sum():,}")
````

---
## Ensemble Decision Fusion (v3: weighted soft score + Nelder-Mead calibration)

**v3 changes:**
- Nelder-Mead weight optimisation on val F1 (was SLSQP on AUC — F1 better for ICS)
- Calibrated ENS_THRESHOLD on val F1 sweep
- IF/AE/XGB/CNN all contribute as soft scores (no majority vote)


````python
#cell 15
print("--- Ensemble Fusion (Test Set) ---\n")

n_cnn  = len(test_cnn_labels)
OFF    = SEQ_LEN

if_al  = iso_test[OFF:OFF+n_cnn].astype(np.float32)      # IF score
ae_al  = ae_score_test_n[OFF:OFF+n_cnn].astype(np.float32) # AE recon error
xgb_al = xgb_prob_test[OFF:OFF+n_cnn].astype(np.float32) # XGB prob
cnn_al = test_cnn_probs.astype(np.float32)                # CNN prob
y_ens  = test_cnn_labels

# Validation scores for weight optimisation
n_v    = min(len(iso_val), len(xgb_prob_val), len(val_cnn_probs), len(ae_score_val_n)) - OFF
if_v   = iso_val[OFF:OFF+n_v].astype(np.float32)
ae_v   = ae_score_val_n[OFF:OFF+n_v].astype(np.float32)
xgb_v  = xgb_prob_val[OFF:OFF+n_v].astype(np.float32)
cnn_v  = val_cnn_probs[:n_v].astype(np.float32)
y_v    = val_cnn_labels[:n_v]


# v5 FIX: If AE is effectively dead, exclude it from ensemble
ae_standalone_preds = (ae_v > AE_THRESHOLD).astype(np.int8)
ae_standalone_f1 = f1_score(y_v, ae_standalone_preds, zero_division=0)
AE_IS_DEAD = ae_standalone_f1 < 0.20
if AE_IS_DEAD:
    print(f"[v5 WARN] AE standalone F1={ae_standalone_f1:.4f} < 0.20 — AE will be excluded from ensemble")

def neg_f1_4w(w):
    """Negative F1 for 4-component ensemble (IF, AE, XGB, CNN)."""
    w = np.clip(w, 0.01, 0.80)
    w = w / (w.sum() + 1e-9)
    sc   = w[0]*if_v + w[1]*ae_v + w[2]*xgb_v + w[3]*cnn_v
    best = max(f1_score(y_v, (sc > t).astype(int), zero_division=0)
               for t in np.linspace(0.15, 0.85, 50))
    return -best


res = minimize(neg_f1_4w, [0.10, 0.10, 0.55, 0.25],
               method="Nelder-Mead",
               options={"maxiter": 1000, "xatol": 0.005, "fatol": 0.005})
w_opt   = np.clip(res.x, 0.01, 0.80)
w_opt  /= w_opt.sum()
W_IF, W_AE, W_XGB, W_CNN = w_opt

# v5 FIX: Zero out AE weight if AE is dead, re-normalize remaining weights
if AE_IS_DEAD:
    W_AE = 0.0
    w_sum = W_IF + W_XGB + W_CNN
    W_IF  = W_IF / w_sum
    W_XGB = W_XGB / w_sum
    W_CNN = W_CNN / w_sum
    print(f"[v5] AE excluded — Re-normalised weights: IF={W_IF:.3f}  AE=0.000  XGB={W_XGB:.3f}  CNN={W_CNN:.3f}")
else:
    print(f"Optimised weights: IF={W_IF:.3f}  AE={W_AE:.3f}  XGB={W_XGB:.3f}  CNN={W_CNN:.3f}")

# Calibrate ensemble threshold on val F1
ens_val_sc  = W_IF*if_v + W_AE*ae_v + W_XGB*xgb_v + W_CNN*cnn_v
ens_thrs    = np.linspace(0.15, 0.85, 400)
ens_f1s     = [f1_score(y_v, (ens_val_sc > t).astype(int), zero_division=0)
                for t in ens_thrs]
ENS_THRESHOLD = float(ens_thrs[np.argmax(ens_f1s)])
print(f"Calibrated ensemble threshold: {ENS_THRESHOLD:.3f}")

# Test set predictions
ens_sc   = (W_IF*if_al + W_AE*ae_al + W_XGB*xgb_al + W_CNN*cnn_al).astype(np.float32)
ens_pred = (ens_sc > ENS_THRESHOLD).astype(np.int8)

ens_f1   = f1_score(y_ens, ens_pred)
ens_auc  = roc_auc_score(y_ens, ens_sc) if y_ens.sum() > 0 else 0.0
ens_fpr  = ((ens_pred==1) & (y_ens==0)).sum() / max((y_ens==0).sum(), 1)
ens_fnr  = ((ens_pred==0) & (y_ens==1)).sum() / max((y_ens==1).sum(), 1)

print(f"\nEnsemble Test Results:")
print(f"  F1   : {ens_f1:.4f}  {'[OK]' if ens_f1>0.88 else '[BELOW TARGET]'}")
print(f"  AUC  : {ens_auc:.4f}")
print(f"  FPR  : {ens_fpr*100:.2f}%")
print(f"  FNR  : {ens_fnr*100:.2f}%  ← minimize for ICS safety")
print()

# Per-layer F1 summary
l1b  = (if_al  > IF_THRESH).astype(np.int8)
l1bb = (ae_al  > AE_THRESHOLD).astype(np.int8)
l2b  = xgb_pred_test[OFF:OFF+n_cnn].astype(np.int8)
l3b  = (cnn_al > CNN_THRESHOLD).astype(np.int8)

for lbl, pred in [("IF (1a)",l1b),("AE (1b)",l1bb),
                   ("XGBoost",l2b),("CNN-BiLSTM",l3b),("Ensemble",ens_pred)]:
    print(f"  {lbl:<12s}: F1={f1_score(y_ens, pred, zero_division=0):.4f}")

print("\nEnsemble Classification Report:")
print(classification_report(y_ens, ens_pred, target_names=["Normal","Attack"]))
````

---
## Per-Attack-Type Evaluation (critical for ICS safety)

````python
#cell 16
print("Per-Attack-Type F1 (Test Set Ensemble)")
print("="*72)
print("  [v4 FIX] Precision now computed per-type (not global), eval_thresh=0.05")
print()

# v4 FIX: Use a lower threshold for per-type evaluation so short attacks
# (like replay, slow_ramp) are not filtered out by the majority_thresh
EVAL_THRESH = 0.05  # at least 1 step in 150 must be the target attack type

results = []
for aid, aname in sorted(ATTACK_TYPE_MAP.items()):
    if aid == 0: continue

    # Build per-type ground truth: 1 if this window contains ≥EVAL_THRESH of target attack
    type_mask = np.array([
        int((y_type_test[OFF+i:OFF+i+SEQ_LEN] == aid).mean() >= EVAL_THRESH)
        for i in range(n_cnn)
    ], dtype=np.int8)
    n_type = type_mask.sum()
    if n_type == 0:
        print(f"  {aname:<24s} — no windows found (EVAL_THRESH={EVAL_THRESH}, SEQ_LEN={SEQ_LEN})")
        continue

    # v4 FIX: Per-type precision using sklearn — denominator is per-type predictions
    from sklearn.metrics import precision_score, recall_score, f1_score as f1s
    y_true_t = type_mask
    y_pred_t = ens_pred.astype(np.int8)

    prec_t  = precision_score(y_true_t, y_pred_t, zero_division=0)
    rec_t   = recall_score(y_true_t, y_pred_t, zero_division=0)
    f1_t    = f1s(y_true_t, y_pred_t, zero_division=0)

    flag = "[CRITICAL MISS]" if rec_t < 0.50 else ("[OK]" if rec_t > 0.80 else "[WARN]")
    print(f"  {aname:<24s} n={n_type:6,} | Recall={rec_t:.3f} | P={prec_t:.3f} | F1={f1_t:.3f}  {flag}")
    results.append((aname, n_type, rec_t, f1_t))

if results:
    worst = min(results, key=lambda x: x[2])
    best  = max(results, key=lambda x: x[2])
    print(f"\nOverall ensemble F1 : {ens_f1:.4f}")
    print(f"Worst attack type   : {worst[0]} (recall={worst[2]:.3f})")
    print(f"Best  attack type   : {best[0]}  (recall={best[2]:.3f})")
    # v4: Warn if any recall < 0.50
    critical_misses = [r for r in results if r[2] < 0.50]
    if critical_misses:
        print(f"\n[ALERT] {len(critical_misses)} CRITICAL MISS(ES) — "
              f"attacks not detected at 50% recall threshold:")
        for cm in critical_misses:
            print(f"  → {cm[0]} (recall={cm[2]:.3f}, n={cm[1]:,})")
    else:
        print("\n[OK] All attack types detected at ≥50% recall.")
````
