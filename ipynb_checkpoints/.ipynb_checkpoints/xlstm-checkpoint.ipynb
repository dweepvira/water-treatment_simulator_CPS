```python
# CELL 1 — INSTALL & IMPORTS
!pip install torch torchvision torchaudio -q

import torch
import torch.nn as nn
import numpy as np
import pandas as pd
from torch.utils.data import Dataset, DataLoader
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import mean_squared_error, r2_score
import matplotlib.pyplot as plt
from tqdm import tqdm
import warnings
warnings.filterwarnings('ignore')

DEVICE = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print(f"Device  : {DEVICE}")
if torch.cuda.is_available():
    print(f"GPU     : {torch.cuda.get_device_name(0)}")
    print(f"VRAM    : {torch.cuda.get_device_properties(0).total_memory/1e9:.1f} GB")
```

```python
# %%
# CELL 3 — DEFINE FEATURE COLUMNS
# Adjust if your column names differ
DYNAMIC_FEATURES = [
    'prcp', 'tmax', 'tmin', 'tavg',
    'srad_lw', 'srad_sw',
    'wind_u', 'wind_v', 'wind',
    'rel_hum',
    'pet', 'pet_gleam', 'aet_gleam',
    'evap_canopy', 'evap_surface',
    'sm_lvl1', 'sm_lvl2', 'sm_lvl3', 'sm_lvl4'
]

STATIC_FEATURES = [
    'elev_mean', 'slope_mean', 'cwc_area', 'dpsbar',
    'p_mean', 'pet_mean', 'aridity_p_pet', 'p_annual_variability',
    'high_prec_freq', 'low_prec_freq', 'asynchronicity',
    'soil_depth', 'soil_conductivity_top', 'soil_awc_top',
    'sand_frac_top', 'clay_frac_top',
    'geol_porosity', 'geol_permeability',
    'crops_frac', 'trees_frac', 'lai_mean',
    'num_dams', 'reservoir_index', 'pop_density_2020'
]

TARGET = 'streamflow_m3s'

# Verify
missing = [c for c in DYNAMIC_FEATURES + STATIC_FEATURES + [TARGET] if c not in df.columns]
print(f"Missing cols : {missing if missing else 'None — all good'}")
print(f"Dynamic      : {len(DYNAMIC_FEATURES)}")
print(f"Static       : {len(STATIC_FEATURES)}")
print(f"Target       : {TARGET}")
```

```python
# CELL 4 — LOG TRANSFORM + SPLIT
# Log-transform target
df['target_log'] = np.log1p(df[TARGET])
TARGET_LOG = 'target_log'

# Time-based split
train_df = df[df['date'] <= '2005-12-31'].copy()
val_df   = df[(df['date'] > '2005-12-31') & (df['date'] <= '2012-12-31')].copy()
test_df  = df[df['date'] > '2012-12-31'].copy()

print(f"Train : {len(train_df):>10,} rows")
print(f"Val   : {len(val_df):>10,} rows")
print(f"Test  : {len(test_df):>10,} rows")
```

```python
# CELL 5 — NORMALISE
# Fill NaN
for col in DYNAMIC_FEATURES:
    med = train_df[col].median()
    for split in [train_df, val_df, test_df]:
        split[col] = split[col].fillna(med)

for col in STATIC_FEATURES:
    med = train_df[col].median()
    for split in [train_df, val_df, test_df]:
        split[col] = split[col].fillna(med)

# Fit scalers on train only
dyn_scaler    = StandardScaler()
static_scaler = StandardScaler()

train_df[DYNAMIC_FEATURES] = dyn_scaler.fit_transform(train_df[DYNAMIC_FEATURES])
train_df[STATIC_FEATURES]  = static_scaler.fit_transform(train_df[STATIC_FEATURES])

val_df[DYNAMIC_FEATURES]   = dyn_scaler.transform(val_df[DYNAMIC_FEATURES])
val_df[STATIC_FEATURES]    = static_scaler.transform(val_df[STATIC_FEATURES])

test_df[DYNAMIC_FEATURES]  = dyn_scaler.transform(test_df[DYNAMIC_FEATURES])
test_df[STATIC_FEATURES]   = static_scaler.transform(test_df[STATIC_FEATURES])

print("Normalisation done")
```

```python
# # Cell 6 — Dataset & DataLoader

# %%
SEQ_LEN    = 30
BATCH_SIZE = 128

class CamelsDataset(Dataset):
    def __init__(self, df, dyn_cols, static_cols, target_col, seq_len):
        self.samples = []
        for gauge_id, grp in df.groupby('gauge_id'):
            grp = grp.sort_values('date').reset_index(drop=True)
            if len(grp) <= seq_len:
                continue
            dyn    = grp[dyn_cols].values.astype(np.float32)
            static = grp[static_cols].iloc[0].values.astype(np.float32)
            target = grp[target_col].values.astype(np.float32)
            for i in range(seq_len, len(grp)):
                self.samples.append((
                    dyn[i - seq_len : i],
                    static,
                    target[i]
                ))

    def __len__(self):
        return len(self.samples)

    def __getitem__(self, idx):
        x_d, x_s, y = self.samples[idx]
        return (
            torch.tensor(x_d, dtype=torch.float32),
            torch.tensor(x_s, dtype=torch.float32),
            torch.tensor(y,   dtype=torch.float32)
        )

print("Building datasets...")
train_ds = CamelsDataset(train_df, DYNAMIC_FEATURES, STATIC_FEATURES, TARGET_LOG, SEQ_LEN)
val_ds   = CamelsDataset(val_df,   DYNAMIC_FEATURES, STATIC_FEATURES, TARGET_LOG, SEQ_LEN)
test_ds  = CamelsDataset(test_df,  DYNAMIC_FEATURES, STATIC_FEATURES, TARGET_LOG, SEQ_LEN)

# Windows-safe dataloaders (num_workers removed to prevent deadlocks)
train_loader = DataLoader(train_ds, batch_size=BATCH_SIZE, shuffle=True,  pin_memory=True)
val_loader   = DataLoader(val_ds,   batch_size=BATCH_SIZE, shuffle=False, pin_memory=True)
test_loader  = DataLoader(test_ds,  batch_size=BATCH_SIZE, shuffle=False, pin_memory=True)
```

```python
# # Cell 7 — xLSTM Model (Stabilized)

# %%
class mLSTMCell(nn.Module):
    def __init__(self, input_dim, hidden_dim, num_heads=4):
        super().__init__()
        self.hidden_dim = hidden_dim
        self.W_i = nn.Linear(input_dim + hidden_dim, hidden_dim)
        self.W_f = nn.Linear(input_dim + hidden_dim, hidden_dim)
        self.W_o = nn.Linear(input_dim + hidden_dim, hidden_dim)
        self.W_q = nn.Linear(input_dim, hidden_dim)
        self.W_k = nn.Linear(input_dim, hidden_dim)
        self.W_v = nn.Linear(input_dim, hidden_dim)
        self.norm = nn.LayerNorm(hidden_dim)

    def forward(self, x, h, C, n):
        combined = torch.cat([x, h], dim=-1)
        
        # CHANGE: Clamps added to prevent torch.exp() overflow NaNs
        w_i_out = torch.clamp(self.W_i(combined), min=-15.0, max=15.0)
        w_f_out = torch.clamp(self.W_f(combined), min=-15.0, max=15.0)
        
        i_gate   = torch.exp(w_i_out)
        f_gate   = torch.exp(w_f_out)
        o_gate   = torch.sigmoid(self.W_o(combined))

        # CHANGE: + 1e-6 added to all torch.log calls to prevent log(0)
        max_gate = torch.max(
            torch.log(f_gate + 1e-6) + torch.log(n + 1e-6),
            torch.log(i_gate + 1e-6)
        ).detach()
        
        i_s = torch.exp(torch.log(i_gate + 1e-6) - max_gate)
        f_s = torch.exp(torch.log(f_gate + 1e-6) + torch.log(n + 1e-6) - max_gate)

        q = self.W_q(x) / (self.hidden_dim ** 0.5)
        k = self.W_k(x) / (self.hidden_dim ** 0.5)
        v = self.W_v(x)

        vk    = torch.bmm(v.unsqueeze(2), k.unsqueeze(1))
        C_new = f_s.unsqueeze(2) * C + i_s.unsqueeze(2) * vk
        n_new = f_s * n + i_s

        h_tilde = torch.bmm(C_new, q.unsqueeze(2)).squeeze(2)
        denom   = torch.clamp(
            torch.abs(torch.bmm(n_new.unsqueeze(1), q.unsqueeze(2)).squeeze()),
            min=1.0
        ).unsqueeze(1)
        h_tilde = h_tilde / denom
        h_new   = o_gate * torch.tanh(self.norm(h_tilde))

        return h_new, C_new, n_new

class HydroXLSTM(nn.Module):
    def __init__(self, n_dynamic, n_static, hidden_dim=128,
                 num_layers=2, num_heads=4, dropout=0.2):
        super().__init__()
        self.hidden_dim = hidden_dim
        self.num_layers = num_layers
        self.input_proj = nn.Linear(n_dynamic, hidden_dim)
        self.mlstm_cells = nn.ModuleList([
            mLSTMCell(hidden_dim, hidden_dim, num_heads)
            for _ in range(num_layers)
        ])
        self.dropout    = nn.Dropout(dropout)
        self.norm       = nn.LayerNorm(hidden_dim)
        self.static_enc = nn.Sequential(
            nn.Linear(n_static, hidden_dim),
            nn.ReLU(),
            nn.Dropout(dropout)
        )
        self.head = nn.Sequential(
            nn.Linear(hidden_dim * 2, hidden_dim),
            nn.ReLU(),
            nn.Dropout(dropout),
            nn.Linear(hidden_dim, 1)
        )

    def forward(self, x_dyn, x_static):
        batch = x_dyn.size(0)
        x     = self.input_proj(x_dyn)

        h_list = [torch.zeros(batch, self.hidden_dim).to(x.device) for _ in range(self.num_layers)]
        C_list = [torch.zeros(batch, self.hidden_dim, self.hidden_dim).to(x.device) for _ in range(self.num_layers)]
        n_list = [torch.ones(batch,  self.hidden_dim).to(x.device) for _ in range(self.num_layers)]

        for t in range(x.size(1)):
            x_t = x[:, t, :]
            for li, cell in enumerate(self.mlstm_cells):
                h_new, C_new, n_new = cell(x_t, h_list[li], C_list[li], n_list[li])
                h_list[li] = self.dropout(h_new)
                C_list[li] = C_new
                n_list[li] = n_new
                x_t = h_new

        h_last   = self.norm(h_list[-1])
        s_enc    = self.static_enc(x_static)
        combined = torch.cat([h_last, s_enc], dim=1)
        return self.head(combined).squeeze(-1)


N_DYNAMIC = len(DYNAMIC_FEATURES)
N_STATIC  = len(STATIC_FEATURES)

model = HydroXLSTM(
    n_dynamic  = N_DYNAMIC,
    n_static   = N_STATIC,
    hidden_dim = 128,
    num_layers = 2,
    num_heads  = 4,
    dropout    = 0.2
).to(DEVICE)

total_params = sum(p.numel() for p in model.parameters() if p.requires_grad)
print(f"Model on       : {next(model.parameters()).device}")
print(f"Total params   : {total_params:,}")

# %% [markdown]
```

```python
# # Cell 8 — Training (bfloat16)

# %%
EPOCHS = 28
LR     = 1e-3

# Fused Adam: executes entirely on the GPU in a single kernel step
optimizer = torch.optim.Adam(model.parameters(), lr=LR, fused=True)
scheduler = torch.optim.lr_scheduler.ReduceLROnPlateau(
    optimizer, mode='min', factor=0.5, patience=5
)
criterion = nn.MSELoss()

# CHANGE: GradScaler is removed entirely. bfloat16 doesn't need it.

train_losses, val_losses = [], []
best_val_loss    = float('inf')
best_model_state = None

print(f"Training  : {EPOCHS} epochs on {DEVICE}")
print("=" * 55)

for epoch in range(1, EPOCHS + 1):

    # Train
    model.train()
    batch_losses = []
    
    train_pbar = tqdm(train_loader, desc=f"Epoch {epoch:3d}/{EPOCHS} [Train]")
    
    for x_dyn, x_static, y in train_pbar:
        x_dyn = x_dyn.to(DEVICE, non_blocking=True)
        x_static = x_static.to(DEVICE, non_blocking=True)
        y = y.to(DEVICE, non_blocking=True)
        
        optimizer.zero_grad(set_to_none=True)
        
        # CHANGE: Switched from float16 to bfloat16
        with torch.autocast(device_type='cuda', dtype=torch.bfloat16):
            pred = model(x_dyn, x_static)
            loss = criterion(pred, y)
            
        # CHANGE: Standard backward pass without scaler
        loss.backward()
        nn.utils.clip_grad_norm_(model.parameters(), max_norm=1.0)
        optimizer.step()
        
        batch_losses.append(loss.item())
        train_pbar.set_postfix({'loss': f"{loss.item():.4f}"})
        
    train_loss = np.mean(batch_losses)

    # Validate
    model.eval()
    val_batch_losses = []
    with torch.no_grad():
        for x_dyn, x_static, y in val_loader:
            x_dyn = x_dyn.to(DEVICE, non_blocking=True)
            x_static = x_static.to(DEVICE, non_blocking=True)
            y = y.to(DEVICE, non_blocking=True)
            
            # CHANGE: Switched from float16 to bfloat16
            with torch.autocast(device_type='cuda', dtype=torch.bfloat16):
                pred = model(x_dyn, x_static)
                loss = criterion(pred, y)
            val_batch_losses.append(loss.item())
    val_loss = np.mean(val_batch_losses)

    train_losses.append(train_loss)
    val_losses.append(val_loss)
    scheduler.step(val_loss)

    if val_loss < best_val_loss:
        best_val_loss    = val_loss
        best_model_state = {k: v.cpu().clone() for k, v in model.state_dict().items()}

    print(f"End of Epoch {epoch} | Train Loss: {train_loss:.5f} | Val Loss: {val_loss:.5f} | LR: {optimizer.param_groups[0]['lr']:.6f}\n")

model.load_state_dict(best_model_state)
print(f"Done. Best val loss : {best_val_loss:.5f}")
```

```python
# %%
# CELL 10 — SAVE MODEL AND SCALERS
import torch
import pickle
import os

# Create a directory to store your saved files
save_dir = "saved_models"
os.makedirs(save_dir, exist_ok=True)

# 1. Save the PyTorch model weights
# We save the 'state_dict' (the actual learned weights) rather than the whole object
model_save_path = os.path.join(save_dir, "xlstm_hydro_weights.pth")
torch.save(model.state_dict(), model_save_path)
print(f"Model weights successfully saved to:\n -> {model_save_path}")

# 2. Save the scalers (CRITICAL for making future predictions)
scalers_save_path = os.path.join(save_dir, "xlstm_scalers.pkl")
with open(scalers_save_path, 'wb') as f:
    pickle.dump({
        'dyn_scaler': dyn_scaler,
        'static_scaler': static_scaler
    }, f)
print(f"Data scalers successfully saved to:\n -> {scalers_save_path}")
```

```python
# CELL 9 — EVALUATE + PLOT
def nse(obs, sim):
    return 1 - np.sum((obs-sim)**2) / np.sum((obs-np.mean(obs))**2)

def kge(obs, sim):
    r = np.corrcoef(obs, sim)[0,1]
    # Added + 1e-8 to prevent divide-by-zero errors
    a = np.std(sim)  / (np.std(obs) + 1e-8)
    b = np.mean(sim) / (np.mean(obs) + 1e-8)
    return 1 - np.sqrt((r-1)**2 + (a-1)**2 + (b-1)**2)

# Inference
model.eval()
preds, trues = [], []
with torch.no_grad():
    for x_dyn, x_static, y in tqdm(test_loader, desc='Test inference'):
        # Keep non_blocking=True for fast data transfer
        x_dyn = x_dyn.to(DEVICE, non_blocking=True)
        x_static = x_static.to(DEVICE, non_blocking=True)
        
        # Fast bfloat16 inference
        with torch.autocast(device_type='cuda', dtype=torch.bfloat16):
            pred = model(x_dyn, x_static)
            
        # THE FIX: Cast to .float() before calling .numpy()
        preds.extend(pred.float().cpu().numpy())
        trues.extend(y.numpy())

preds = np.array(preds)
trues = np.array(trues)

# Back-transform
obs  = np.expm1(trues)
pred = np.clip(np.expm1(preds), 0, None)

rmse_ = np.sqrt(mean_squared_error(obs, pred))
r2_   = r2_score(obs, pred)
nse_  = nse(obs, pred)
kge_  = kge(obs, pred)

print("=" * 50)
print("  TEST SET RESULTS")
print("=" * 50)
print(f"  NSE  : {nse_:.4f}")
print(f"  KGE  : {kge_:.4f}")
print(f"  RMSE : {rmse_:.3f} m3/s")
print(f"  R2   : {r2_:.4f}")
print("=" * 50)

# Learning curves
fig, ax = plt.subplots(figsize=(10,4))
ax.plot(train_losses, label='Train', color='steelblue')
ax.plot(val_losses,   label='Val',   color='tomato')
ax.set_xlabel('Epoch')
ax.set_ylabel('MSE Loss')
ax.set_yscale('log')
ax.legend()
ax.set_title('xLSTM Training Curves')
plt.tight_layout()
plt.show()
```

```python
# CELL 9 — EVALUATE + PLOT
import matplotlib.pyplot as plt
from tqdm import tqdm
from sklearn.metrics import mean_squared_error, r2_score

def nse(obs, sim):
    return 1 - np.sum((obs-sim)**2) / np.sum((obs-np.mean(obs))**2)

def kge(obs, sim):
    r = np.corrcoef(obs, sim)[0,1]
    a = np.std(sim)  / (np.std(obs) + 1e-8)
    b = np.mean(sim) / (np.mean(obs) + 1e-8)
    return 1 - np.sqrt((r-1)**2 + (a-1)**2 + (b-1)**2)

# Inference
model.eval()
preds, trues = [], []
with torch.no_grad():
    for x_dyn, x_static, y in tqdm(test_loader, desc='Test inference'):
        x_dyn = x_dyn.to(DEVICE, non_blocking=True)
        x_static = x_static.to(DEVICE, non_blocking=True)
        
        # Fast bfloat16 inference
        with torch.autocast(device_type='cuda', dtype=torch.bfloat16):
            pred = model(x_dyn, x_static)
            
        # THE FIX: Cast to .float() before calling .numpy()
        preds.extend(pred.float().cpu().numpy())
        trues.extend(y.numpy())

preds = np.array(preds)
trues = np.array(trues)

# Back-transform from log scale to normal m3/s
obs  = np.expm1(trues)
pred = np.clip(np.expm1(preds), 0, None)

# Calculate Metrics
rmse_ = np.sqrt(mean_squared_error(obs, pred))
r2_   = r2_score(obs, pred)
nse_  = nse(obs, pred)
kge_  = kge(obs, pred)

print("=" * 50)
print("  TEST SET RESULTS (2013 Onwards)")
print("=" * 50)
print(f"  NSE  : {nse_:.4f}")
print(f"  KGE  : {kge_:.4f}")
print(f"  RMSE : {rmse_:.3f} m3/s")
print(f"  R2   : {r2_:.4f}")
print("=" * 50)

# Learning curves
fig, ax = plt.subplots(figsize=(10,4))
ax.plot(train_losses, label='Train', color='steelblue')
ax.plot(val_losses,   label='Val',   color='tomato')
ax.set_xlabel('Epoch')
ax.set_ylabel('MSE Loss')
ax.set_yscale('log')
ax.legend()
ax.set_title('xLSTM Training Curves')
plt.tight_layout()
plt.show()
```
